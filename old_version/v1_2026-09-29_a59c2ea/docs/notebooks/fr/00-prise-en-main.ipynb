## 2. Variables et types
Une **variable** est un nom qui désigne une valeur. Le signe `=` signifie « reçoit », et non « est égal à » : `annee = 1956` range 1956 sous le nom `annee`. Chaque valeur a un **type** :
- `int` : un nombre entier (`1956`) ;
- `float` : un nombre décimal, écrit avec un point (`250.0`) ;
- `str` : une chaîne de caractères, c'est-à-dire du texte entre guillemets (`"Nedjma"`) ;
- `bool` : un booléen, qui vaut `True` (vrai) ou `False` (faux).

Cette cellule décrit un roman par une fiche de lecture de cinq variables. Le nombre de mots est une valeur arrondie, fictive. `type()` affiche le type d'une valeur ; `<` signifie « est plus petit que ».

**Avant d'exécuter :** combien de pages fait un roman de 95 000 mots, à 250 mots par page ? Ce roman a-t-il été publié avant 1962 ?

````python
#@title À MODIFIER : une fiche de lecture
titre = "Nedjma"  #@param {type:"string"}
annee = 1956  #@param {type:"integer"}
nb_mots = 95000  #@param {type:"integer"}
mots_par_page = 250.0  #@param {type:"number"}
deja_lu = False  #@param {type:"boolean"}

print(type(titre), type(annee), type(mots_par_page), type(deja_lu))
print("Pages estimées :", nb_mots / mots_par_page)   # / : la division
print("Publié avant 1962 ?", annee < 1962)           # une comparaison : vrai ou faux
````

### Ce que vous devez voir
```
<class 'str'> <class 'int'> <class 'float'> <class 'bool'>
Pages estimées : 380.0
Publié avant 1962 ? True
```
- La division `/` donne toujours un nombre décimal : `380.0`.
- La comparaison `annee < 1962` donne un booléen : `True`.
- Les guillemets de l'étape 1 signalaient donc une chaîne de caractères (`str`), et `1929` était un nombre entier (`int`).

## 3. D'une chaîne de caractères à une liste de mots
Un texte, pour Python, est une **chaîne de caractères**. Les chaînes ont des **méthodes**, des actions que l'on appelle avec un point :
- `lower()` met tout en minuscules ;
- `replace(a, b)` remplace `a` par `b` (ici, on remplace la ponctuation par rien) ;
- `split()` découpe aux espaces et renvoie une **liste** de mots.

`len()` donne une longueur : le nombre de caractères d'une chaîne, ou le nombre d'éléments d'une liste. Dans une liste, `[0]` désigne le premier élément et `[-1]` le dernier.

Le vers vient de Baudelaire, « L'Invitation au voyage » (*Les Fleurs du mal*, 1857).

**Avant d'exécuter :** combien de mots compte ce vers ? Comptez sur vos doigts.

````python
#@title Découper un vers en mots
vers = "Mon enfant, ma sœur, / Songe à la douceur / D'aller là-bas vivre ensemble !"
print(len(vers))                  # nombre de caractères, espaces compris
print(vers.lower())               # tout en minuscules
# Retirer la ponctuation, puis découper aux espaces
mots_du_vers = vers.replace(",", "").replace(" /", "").replace(" !", "").split()
print(mots_du_vers)
print(len(mots_du_vers), "mots ; premier :", mots_du_vers[0], "; dernier :", mots_du_vers[-1])
````

### Ce que vous devez voir
```

['Mon', 'enfant', 'ma', 'sœur', 'Songe', 'à', 'la', 'douceur', "D'aller", 'là-bas', 'vivre', 'ensemble']
12 mots ; premier : Mon ; dernier : ensemble
```
La liste s'écrit entre crochets `[ ]`, chaque mot entre guillemets. `vers` n'a pas changé : `lower()` et `replace()` fabriquent une nouvelle chaîne.

**Observez :** pour Python, « D'aller » et « là-bas » sont chacun **un** mot. Et pour vous ? Et pour un dictionnaire de la langue française ?

## 4. Les listes
Une **liste** est une suite ordonnée de valeurs, écrite entre crochets. On y ajoute un élément à la fin avec `append()`, on la trie avec `sort()`, et on lit ses éléments par leur position, appelée **indice**, qui commence à 0. `auteurs[1:3]` est une **tranche** : les éléments d'indice 1 et 2.

**Avant d'exécuter :** après l'ajout, combien la liste compte-t-elle d'auteurs ? Lequel sera le premier une fois la liste triée ?

````python
#@title À MODIFIER : une liste d'auteurs
nouvel_auteur = "Kateb Yacine"  #@param {type:"string"}

auteurs = ["Hugo", "Baudelaire", "Maupassant"]   # une liste de 3 chaînes
auteurs.append(nouvel_auteur)                     # ajouter un élément à la fin
print(auteurs[0], "|", auteurs[-1])               # le premier et le dernier élément
print(len(auteurs), "auteurs")
auteurs.sort()                                    # trier la liste
print(auteurs)
print(auteurs[1:3])                               # une tranche : indices 1 et 2
````

### Ce que vous devez voir
```

4 auteurs
['Baudelaire', 'Hugo', 'Kateb Yacine', 'Maupassant']
['Hugo', 'Kateb Yacine']
```
Le tri suit l'ordre alphabétique. La tranche `[1:3]` s'arrête **avant** l'indice 3.

## 5. Les dictionnaires
Un **dictionnaire** associe des **clés** à des **valeurs**, comme une fiche de lecture associe des rubriques à leur contenu. Il s'écrit entre accolades `{ }`, avec des paires `clé: valeur`. On lit une valeur par sa clé : `fiche["auteur"]`. `get(clé, valeur_de_secours)` renvoie la valeur de secours quand la clé n'existe pas, au lieu de s'arrêter sur une erreur.

**Avant d'exécuter :** que va afficher la première ligne ? La fiche contient-elle une rubrique « genre » ?

````python
#@title À MODIFIER : une fiche de lecture en dictionnaire
rubrique = "genre"  #@param ["genre", "auteur", "annee", "langue", "titre_fr"]

fiche = {
    "titre": "Bayn al-Qasrayn",
    "auteur": "Naguib Mahfouz",
    "annee": 1956,
    "langue": "arabe",
}
print(fiche["auteur"])                             # lire une valeur par sa clé
fiche["titre_fr"] = "Impasse des deux palais"      # ajouter une nouvelle paire
print(fiche)
print(len(fiche), "rubriques")
print(rubrique, "->", fiche.get(rubrique, "non renseigné"))   # valeur de secours si la clé manque
````

### Ce que vous devez voir
```
Naguib Mahfouz
{'titre': 'Bayn al-Qasrayn', 'auteur': 'Naguib Mahfouz', 'annee': 1956, 'langue': 'arabe', 'titre_fr': 'Impasse des deux palais'}
5 rubriques
genre -> non renseigné
```
La clé `"titre_fr"` a été ajoutée à la fin. La rubrique « genre » n'existe pas : `get()` renvoie la valeur de secours.

## 6. Une boucle
Une **boucle** `for` répète des instructions pour chaque élément d'une liste. Les lignes répétées sont **décalées** de quatre espaces sous la ligne `for ... :` ; la première ligne non décalée n'est exécutée qu'une fois, après la boucle. Cette cellule reprend la liste `auteurs` de l'étape 4 et la fiche de l'étape 5.

**Avant d'exécuter :** combien de lignes la première boucle va-t-elle afficher ? Combien de caractères compte « Kateb Yacine » ?

````python
#@title Parcourir une liste, puis un dictionnaire, avec une boucle
total = 0
for auteur in auteurs:                   # pour chaque auteur de la liste...
    print(auteur, ":", len(auteur), "caractères")   # ...ces lignes décalées sont répétées
    total = total + len(auteur)          # on ajoute la longueur au total
print("Total :", total, "caractères pour", len(auteurs), "auteurs")   # après la boucle
print("---")
for cle, valeur in fiche.items():        # items() : les paires clé-valeur
    print(cle, ":", valeur)
````

### Ce que vous devez voir
```
Baudelaire : 10 caractères
Hugo : 4 caractères
Kateb Yacine : 12 caractères
Maupassant : 10 caractères
Total : 36 caractères pour 4 auteurs
---
titre : Bayn al-Qasrayn
auteur : Naguib Mahfouz
annee : 1956
langue : arabe
titre_fr : Impasse des deux palais
```
Les auteurs sont dans l'ordre alphabétique, car la liste a été triée à l'étape 4. « Kateb Yacine » compte 12 caractères : `len()` compte aussi l'espace.

**Observez :** la variable `total` grandit à chaque tour de boucle : 10, puis 14, puis 26, puis 36. Si vous avez changé `nouvel_auteur` à l'étape 4, vos nombres sont différents : c'est normal.

## 7. Un poème à compter
Nous allons compter les mots d'un poème du domaine public : Paul Verlaine, « Il pleure dans mon cœur », *Romances sans paroles* (1874). La première cellule range le poème dans la variable `texte` ; les triples guillemets `"""` permettent une chaîne sur plusieurs lignes.

**Avant d'exécuter :** cette cellule affiche-t-elle quelque chose ?

````python
#@title Le poème : Verlaine, « Il pleure dans mon cœur » (1874)
texte = """Il pleure dans mon cœur
Comme il pleut sur la ville ;
Quelle est cette langueur
Qui pénètre mon cœur ?

Ô bruit doux de la pluie
Par terre et sur les toits !
Pour un cœur qui s'ennuie,
Ô le chant de la pluie !

Il pleure sans raison
Dans ce cœur qui s'écœure.
Quoi ! nulle trahison ?...
Ce deuil est sans raison.

C'est bien la pire peine
De ne savoir pourquoi
Sans amour et sans haine
Mon cœur a tant de peine !"""
````

### Ce que vous devez voir
Rien ! Ranger une valeur dans une variable n'affiche rien. Le poème est maintenant en mémoire, sous le nom `texte`, pour les cellules suivantes.

**Avant d'exécuter la cellule suivante :** le poème compte quatre strophes. Combien de **lignes** (lignes vides comprises), combien de **vers** et environ combien de **mots** ?

La cellule suivante utilise une condition `if` : la ligne décalée sous `if` n'est exécutée que si la condition est vraie. `!=` signifie « différent de ». `re.findall(r"\w+", ...)` trouve toutes les suites de lettres : c'est notre façon de découper en mots en laissant la ponctuation de côté.

````python
#@title Compter les lignes, les vers et les mots
import re                                   # le module des expressions régulières

lignes = texte.split("\n")                  # une chaîne par ligne ("\n" = fin de ligne)
vers = []
for ligne in lignes:
    if ligne.strip() != "":                 # on ignore les lignes vides entre les strophes
        vers.append(ligne)
mots = re.findall(r"\w+", texte.lower())    # les mots, en minuscules, sans ponctuation
print(len(lignes), "lignes,", len(vers), "vers,", len(mots), "mots")
print(mots[:12])                            # les 12 premiers mots
````

### Ce que vous devez voir
```
19 lignes, 16 vers, 82 mots
['il', 'pleure', 'dans', 'mon', 'cœur', 'comme', 'il', 'pleut', 'sur', 'la', 'ville', 'quelle']
```
Les 19 lignes comptent 16 vers et 3 lignes vides entre les strophes. Les mots sont en minuscules et la ponctuation a disparu.

**Observez :** « s'ennuie » devient deux mots, `s` et `ennuie`, et « C'est » devient `c` et `est`. Comment traiter l'élision du français pour compter les mots d'un texte ?

## 8. Compter chaque mot avec une boucle et un dictionnaire
Construisons « à la main » un **dictionnaire de fréquences** : chaque mot (la clé) reçoit son nombre d'occurrences (la valeur). À chaque tour de boucle, `frequences.get(mot, 0)` donne le nombre déjà compté pour ce mot, ou 0 s'il est nouveau, et l'on ajoute 1. C'est l'exercice 3 du TD du chapitre 2.

**Avant d'exécuter :** combien de fois le mot « cœur » apparaît-il dans le poème ? Combien le poème a-t-il de mots **différents** ?

````python
#@title À MODIFIER : compter chaque mot avec une boucle
mot_cherche = "cœur"  #@param {type:"string"}

frequences = {}                              # un dictionnaire vide
for mot in mots:
    # get(mot, 0) : le nombre déjà compté pour ce mot, ou 0 s'il est nouveau
    frequences[mot] = frequences.get(mot, 0) + 1
print(len(mots), "mots,", len(frequences), "mots différents")
print(mot_cherche, ":", frequences.get(mot_cherche.lower(), 0), "fois")
````

### Ce que vous devez voir
```
82 mots, 51 mots différents
cœur : 5 fois
```
Le poème emploie 51 mots différents : les occurrences (82) sont les apparitions des mots, les formes (51) les mots différents.

**Modifiez :** cherchez `pleure`, puis `pleut`, puis `s`. Le programme compte « pleure » (du verbe pleurer) et « pleut » (du verbe pleuvoir) comme deux mots sans rapport : le rapprochement sonore qui fait le poème lui échappe. C'est à vous de l'interpréter.

## 9. Les mots les plus fréquents
`Counter`, du module `collections`, fait le même comptage en une ligne, et `most_common(n)` renvoie les `n` mots les plus fréquents. Les **mots vides** (*stop words*) sont les mots-outils présents dans tous les textes : articles, pronoms, prépositions. On les retire pour voir les mots « pleins ». La liste de mots vides est un choix du chercheur : vous pouvez en ajouter dans le champ `autres_mots_vides`, séparés par des virgules.

**Avant d'exécuter :** parmi les 5 mots les plus fréquents du poème, combien sont des mots pleins (noms, verbes, adjectifs) ? Et une fois les mots vides retirés, quel mot sera en tête ?

````python
#@title À MODIFIER : les mots les plus fréquents
nombre_de_mots = 5  #@param {type:"slider", min:3, max:15, step:1}
autres_mots_vides = ""  #@param {type:"string"}

from collections import Counter              # Counter compte les éléments d'une liste
mots_vides = {"il", "mon", "la", "le", "les", "de", "est", "qui", "dans", "sur", "et",
              "s", "c", "ce", "cette", "un", "a", "ne", "par", "pour", "comme", "ô"}
mots_vides.update(autres_mots_vides.lower().replace(",", " ").split())   # vos ajouts
mots_pleins = [m for m in mots if m not in mots_vides]   # les mots qui ne sont pas vides
print("Tous les mots :", Counter(mots).most_common(nombre_de_mots))
plus_frequents = Counter(mots_pleins).most_common(nombre_de_mots)
print("Sans les mots vides :", plus_frequents)
````

### Ce que vous devez voir
```
Tous les mots : [('cœur', 5), ('la', 4), ('de', 4), ('sans', 4), ('il', 3)]
Sans les mots vides : [('cœur', 5), ('sans', 4), ('pleure', 2), ('pluie', 2), ('raison', 2)]
```
Chaque résultat est une liste de paires `(mot, nombre)`. Sans filtre, « cœur » est le seul mot plein des cinq premiers : les autres sont des mots-outils. Une fois les mots vides retirés, « cœur » reste en tête, suivi de « sans », un mot-outil que nous avons laissé exprès hors de la liste.

**Observez :** « peine » apparaît aussi 2 fois, mais n'est pas dans la liste : à égalité, `Counter` garde l'ordre d'apparition dans le texte. Placez le curseur sur 6 pour le voir. Puis tapez `sans` dans `autres_mots_vides` : faut-il retirer ce mot-outil, qui dit l'absence de raison, d'amour et de haine ?

## 10. Un diagramme en barres
La bibliothèque `matplotlib` dessine des graphiques. `plt.bar(etiquettes, hauteurs)` trace une barre par mot, dont la hauteur est le nombre d'occurrences. Un titre et des noms d'axes ne sont pas décoratifs : sans eux, un graphique est ininterprétable. Le graphique reprend la liste `plus_frequents` de l'étape 9.

**Avant d'exécuter :** combien de barres le graphique va-t-il compter ? Quelle sera la plus haute ?

````python
#@title Dessiner les mots les plus fréquents
import matplotlib.pyplot as plt

etiquettes = []
hauteurs = []
for mot, nombre in plus_frequents:          # chaque élément est une paire (mot, nombre)
    etiquettes.append(mot)
    hauteurs.append(nombre)
plt.figure(figsize=(7, 4))
plt.bar(etiquettes, hauteurs, color="#2a78d6")
plt.yticks(range(0, max(hauteurs) + 1))       # des graduations entières : on compte des mots
plt.title("Verlaine, « Il pleure dans mon cœur » : les mots pleins les plus fréquents")
plt.xlabel("Mot (mots vides retirés)")
plt.ylabel("Nombre d'occurrences")
plt.show()
````

### Ce que vous devez voir
Un diagramme de 5 barres bleues : « cœur » (5), « sans » (4), puis « pleure », « pluie » et « raison » (2 chacun).

**Observez :** le graphique ne montre rien que la liste de l'étape 9 ne contenait déjà. Qu'apporte-t-il à un lecteur pressé ? Qu'est-ce qu'il cache (le contexte de chaque mot, l'ordre des strophes) ?
