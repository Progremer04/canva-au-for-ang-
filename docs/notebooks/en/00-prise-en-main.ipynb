## 2. Variables and types
A **variable** is a name that refers to a value. The `=` sign means "receives", not "is equal to": `year = 1818` stores 1818 under the name `year`. Each value has a **type**:
- `int`: a whole number (`1818`);
- `float`: a decimal number, written with a point (`250.0`);
- `str`: a character string, that is, text in quotation marks (`"Frankenstein"`);
- `bool`: a Boolean, which is either `True` or `False`.

This cell describes a novel with a reading card made of five variables. The number of words is a rounded, made-up value. `type()` displays the type of a value; `<` means "is smaller than".

**Before running:** how many pages does a novel of 75,000 words have, at 250 words per page? Was this novel published before 1837, the start of the Victorian era?

````python
#@title TO CHANGE: a reading card
title = "Frankenstein"  #@param {type:"string"}
year = 1818  #@param {type:"integer"}
nb_words = 75000  #@param {type:"integer"}
words_per_page = 250.0  #@param {type:"number"}
already_read = False  #@param {type:"boolean"}

print(type(title), type(year), type(words_per_page), type(already_read))
print("Estimated pages:", nb_words / words_per_page)   # / : division
print("Published before 1837?", year < 1837)           # a comparison: True or False
````

- Division with `/` always gives a decimal number: `300.0`.
- The comparison `year < 1837` gives a Boolean: `True`.
- So the quotation marks in step 1 marked a character string (`str`), and `1797` was a whole number (`int`).

## 3. From a character string to a list of words
For Python, a text is a **character string**. Strings have **methods**, actions that you call with a dot:
- `lower()` converts everything to lowercase;
- `replace(a, b)` replaces `a` with `b` (here, we replace punctuation with nothing);
- `split()` cuts at the spaces and returns a **list** of words.

`len()` gives a length: the number of characters in a string, or the number of items in a list. In a list, `[0]` refers to the first item and `[-1]` to the last one.

The lines come from William Blake, "The Tyger" (*Songs of Experience*, 1794).

**Before running:** how many words are there in these two lines? Count them on your fingers.

````python
#@title Split a line of poetry into words
lines = "Tyger Tyger, burning bright, / In the forests of the night;"
print(len(lines))                 # number of characters, spaces included
print(lines.lower())              # everything in lowercase
# Remove the punctuation, then cut at the spaces
words_in_lines = lines.replace(",", "").replace(" /", "").replace(";", "").split()
print(words_in_lines)
print(len(words_in_lines), "words, first:", words_in_lines[0], "| last:", words_in_lines[-1])
````

### What you should see
```

['Tyger', 'Tyger', 'burning', 'bright', 'In', 'the', 'forests', 'of', 'the', 'night']
10 words, first: Tyger | last: night
```
The list is written in square brackets `[ ]`, each word in quotation marks. `lines` has not changed: `lower()` and `replace()` make a new string.

**Observe:** the list keeps both "Tyger" and both "the": a list keeps repeated items. "In" starts with a capital letter: will Python treat "In" and "in" as the same word? What does `lower()` do about it?

## 4. Lists
A **list** is an ordered sequence of values, written in square brackets. You add an item at the end with `append()`, sort the list with `sort()`, and read its items by their position, called the **index**, which starts at 0. `authors[1:3]` is a **slice**: the items at index 1 and 2.

**Before running:** after the addition, how many authors does the list contain? Which one will come first once the list is sorted?

````python
#@title TO CHANGE: a list of authors
new_author = "Achebe"  #@param {type:"string"}

authors = ["Shakespeare", "Austen", "Dickens"]    # a list of 3 strings
authors.append(new_author)                        # add an item at the end
print(authors[0], "|", authors[-1])               # the first and the last item
print(len(authors), "authors")
authors.sort()                                    # sort the list
print(authors)
print(authors[1:3])                               # a slice: index 1 and 2
````

## 5. Dictionaries
A **dictionary** links **keys** to **values**, just as a reading card links headings to their content. It is written in curly braces `{ }`, with `key: value` pairs. You read a value by its key: `card["author"]`. `get(key, fallback)` returns the fallback value when the key does not exist, instead of stopping with an error.

**Before running:** what will the first line display? Does the card have a "genre" heading?

````python
#@title TO CHANGE: a reading card as a dictionary
heading = "genre"  #@param ["genre", "author", "year", "language", "title_en"]

card = {
    "title": "Bayn al-Qasrayn",
    "author": "Naguib Mahfouz",
    "year": 1956,
    "language": "Arabic",
}
print(card["author"])                              # read a value by its key
card["title_en"] = "Palace Walk"                   # add a new pair
print(card)
print(len(card), "headings")
print(heading, "->", card.get(heading, "not given"))   # fallback value if the key is missing
````

### What you should see
```
Naguib Mahfouz
{'title': 'Bayn al-Qasrayn', 'author': 'Naguib Mahfouz', 'year': 1956, 'language': 'Arabic', 'title_en': 'Palace Walk'}
5 headings
genre -> not given
```
The key `"title_en"` has been added at the end. The "genre" heading does not exist: `get()` returns the fallback value.

## 6. A loop
A `for` **loop** repeats instructions for each item in a list. The repeated lines are **indented** by four spaces under the `for ... :` line; the first line that is not indented runs only once, after the loop. This cell uses the `authors` list from step 4 and the card from step 5.

**Before running:** how many lines will the first loop display? How many characters are there in "Shakespeare"?

````python
#@title Go through a list, then a dictionary, with a loop
total = 0
for author in authors:                   # for each author in the list...
    print(author, ":", len(author), "characters")   # ...these indented lines are repeated
    total = total + len(author)          # add the length to the total
print("Total:", total, "characters for", len(authors), "authors")   # after the loop
print("---")
for key, value in card.items():          # items(): the key-value pairs
    print(key, ":", value)
````

### What you should see

```
---

author : Naguib Mahfouz
year : 1956
language : Arabic
title_en : Palace Walk
```
The authors are in alphabetical order, because the list was sorted in step 4. "Shakespeare" has 11 characters.

**Observe:** the variable `total` grows at each turn of the loop: 6, then 12, then 19, then 30. If you changed `new_author` in step 4, your numbers are different: that is normal. Type "Chinua Achebe" in step 4 and run steps 4 and 6 again: why does `len()` now give 13?

## 7. A poem to count
We are going to count the words of a public-domain poem: Henry Wadsworth Longfellow, "The Rainy Day", *Ballads and Other Poems* (1841). The first cell stores the poem in the variable `text`; triple quotation marks `"""` allow a string over several lines.

**Before running:** will this cell display anything?

````python
#@title The poem: Longfellow, "The Rainy Day" (1841)
text = """The day is cold, and dark, and dreary;
It rains, and the wind is never weary;
The vine still clings to the mouldering wall,
But at every gust the dead leaves fall,
And the day is dark and dreary.

My life is cold, and dark, and dreary;
It rains, and the wind is never weary;
My thoughts still cling to the mouldering Past,
But the hopes of youth fall thick in the blast,
And the days are dark and dreary.

Be still, sad heart! and cease repining;
Behind the clouds is the sun still shining;
Thy fate is the common fate of all,
Into each life some rain must fall,
Some days must be dark and dreary."""
````

### What you should see
Nothing! Storing a value in a variable displays nothing. The poem is now in memory, under the name `text`, for the next cells.

**Before running the next cell:** the poem has three stanzas. How many **lines** (blank lines included), how many **lines of verse** and roughly how many **words**?

The next cell uses an `if` condition: the indented line under `if` runs only if the condition is true. `!=` means "is different from". `re.findall(r"\w+", ...)` finds every sequence of letters: this is how we cut the text into words while leaving the punctuation aside.

````python
#@title Count the lines, the lines of verse and the words
import re                                   # the regular expressions module

lines = text.split("\n")                    # one string per line ("\n" = end of line)
verse = []
for line in lines:
    if line.strip() != "":                  # skip the blank lines between the stanzas
        verse.append(line)
words = re.findall(r"\w+", text.lower())    # the words, in lowercase, without punctuation
print(len(lines), "lines,", len(verse), "lines of verse,", len(words), "words")
print(words[:12])                           # the first 12 words
````

### What you should see
```
17 lines, 15 lines of verse, 117 words
['the', 'day', 'is', 'cold', 'and', 'dark', 'and', 'dreary', 'it', 'rains', 'and', 'the']
```
The 17 lines contain 15 lines of verse and 2 blank lines between the stanzas. The words are in lowercase and the punctuation has gone.

**Observe:** "The" at the start of the poem has become `the`. Why does this matter when we count how often each word appears?

## 8. Count each word with a loop and a dictionary
Let us build a **frequency dictionary** "by hand": each word (the key) receives its number of occurrences (the value). At each turn of the loop, `frequencies.get(word, 0)` gives the number already counted for this word, or 0 if the word is new, and we add 1. This is exercise 3 of the chapter 2 tutorial.

**Before running:** how many times does the word "dark" appear in the poem? How many **different** words does the poem have?

````python
#@title TO CHANGE: count each word with a loop
search_word = "dark"  #@param {type:"string"}

frequencies = {}                             # an empty dictionary
for word in words:
    # get(word, 0): the number already counted for this word, or 0 if it is new
    frequencies[word] = frequencies.get(word, 0) + 1
print(len(words), "words,", len(frequencies), "different words")
print(search_word, ":", frequencies.get(search_word.lower(), 0), "times")
````

### What you should see
```
117 words, 56 different words
dark : 5 times
```
The poem uses 56 different words: the occurrences (117) are all the appearances of words, the word types (56) are the different words.

**Modify:** search for `rain`, then `rains`, then `still`. The program counts "rain" and "rains" as two unrelated words, and it cannot tell that "still" means "always" in "The vine still clings" but "calm" in "Be still, sad heart!". Interpreting this is your job, not the program's.

## 9. The most frequent words
`Counter`, from the `collections` module, does the same counting in one line, and `most_common(n)` returns the `n` most frequent words. **Stop words** are the function words found in every text: articles, pronouns, prepositions. We remove them to see the content words. The stop-word list is the researcher's choice: you can add some in the `other_stop_words` field, separated by commas.

**Before running:** among the 5 most frequent words in the poem, how many are content words (nouns, verbs, adjectives)? And once the stop words are removed, which word will come first?

````python
#@title TO CHANGE: the most frequent words
number_of_words = 5  #@param {type:"slider", min:3, max:15, step:1}
other_stop_words = ""  #@param {type:"string"}

from collections import Counter              # Counter counts the items in a list
stop_words = {"the", "and", "is", "it", "to", "of", "in", "at", "but", "my", "be", "are",
              "some", "must", "each", "into", "all", "thy", "every", "a"}
stop_words.update(other_stop_words.lower().replace(",", " ").split())   # your additions
content_words = [w for w in words if w not in stop_words]   # the words that are not stop words
print("All words:", Counter(words).most_common(number_of_words))
most_frequent = Counter(content_words).most_common(number_of_words)
print("Without stop words:", most_frequent)
````

### What you should see
```
All words: [('the', 14), ('and', 12), ('is', 7), ('dark', 5), ('dreary', 5)]
Without stop words: [('dark', 5), ('dreary', 5), ('still', 4), ('fall', 3), ('day', 2)]
```
Each result is a list of `(word, number)` pairs. Without a filter, the first three are function words. Once they are removed, "dark" and "dreary" come first.

**Observe:** many other words also appear twice ("cold", "rains", "wind", "weary"…) but only "day" is shown: when there is a tie, `Counter` keeps the order in which the words first appear in the text. Move the slider to 10 to see more of them. Then type `still` in `other_stop_words`: should this word be removed?

## 10. A bar chart
The `matplotlib` library draws charts. `plt.bar(labels, heights)` draws one bar per word, whose height is the number of occurrences. A title and axis labels are not decoration: without them, a chart cannot be interpreted. The chart uses the `most_frequent` list from step 9.

**Before running:** how many bars will the chart have? Which one will be the tallest?

````python
#@title Draw the most frequent words
import matplotlib.pyplot as plt

labels = []
heights = []
for word, number in most_frequent:          # each item is a (word, number) pair
    labels.append(word)
    heights.append(number)
plt.figure(figsize=(7, 4))
plt.bar(labels, heights, color="#2a78d6")
plt.yticks(range(0, max(heights) + 1))        # whole-number ticks: we are counting words
plt.title('Longfellow, "The Rainy Day": the most frequent content words')
plt.xlabel("Word (stop words removed)")
plt.ylabel("Number of occurrences")
plt.show()
````

### What you should see
A chart with 5 blue bars: "dark" and "dreary" (5 each), "still" (4), "fall" (3), then "day" (2).

**Observe:** the chart shows nothing that the list in step 9 did not already contain. What does it offer a reader in a hurry? What does it hide (the context of each word, the order of the stanzas)?
